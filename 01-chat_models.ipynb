```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI

# Carregar variáveis de ambiente
load_dotenv()
chat = ChatOpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key = os.getenv("ROUTER_API"),
    model="gpt-oss-120b:free", 
    temperature=0.7
)
```

### Trabalhando com LLMs e ChatModels no LangChain
- O LangChain fornece uma interface padronizada para conectar diferentes Modelos de Linguagem (LLMs).
- Hoje a forma recomendada é trabalhar com **ChatModels**, que suportam tanto prompts simples quanto interações mais complexas baseadas em mensagens.

### Exemplo 1 - Prompt Simples
- Usando apenas texto, sem mensagens estruturadas

```python
resposta = chat.invoke("Me conte uma piada curta sobre programadores.")
print(resposta.content)
```

### Exemplo 2 - Resposta em Streaming

```python
prompt = "Escreva uma história de aventura em 3 parágrafos, com cliffhangers."
for chunk in chat.stream(prompt):
    print(chunk.content, end="", flush=True)
```

### Exemplo 3 – Batch
- Enviando várias perguntas de uma vez

```python
perguntas = [
    "O que é Memória RAM?",
    "Explique o papel da Memória Cache.",
    "Qual a função do Processador?"
]
respostas = chat.batch(perguntas)

for pergunta, resposta in zip(perguntas, respostas):
    print(f"\n❓ {pergunta}")
    print(f"💡 {resposta.content}")
```

### Exemplo 4 – Chat estruturado
- Usando mensagens de sistema e humano para controlar o estilo da resposta

### Mensagens no LangChain
- SystemMessage define o “papel” ou tom do modelo.
- HumanMessage representa entradas do usuário.
- AIMessage pode ser usado para simular histórico de conversas.

```python
from langchain_core.messages import HumanMessage, SystemMessage
```

```python
mensagens = [
    SystemMessage(content="Você é um professor de tecnologia que responde com humor."),
    HumanMessage(content="Explique a diferença entre HD e SSD.")
]

resposta = chat.invoke(mensagens)
print(resposta.content)
```
