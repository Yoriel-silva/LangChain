```python
import wikipedia
from langchain_core.tools import tool

wikipedia.set_lang("pt")
```

```python
@tool
def busca_wikipedia(query: str):
    """Busca dados no wikipeddia e retorna resumos de páginas para a querry"""
    titulo_paginas = wikipedia.search(query)
    resumos = []
    for titulo in titulo_paginas[:3]:
        try:
            wiki_page = wikipedia.page(title=titulo, auto_suggest=True)
            resumos.append(f"Título: {titulo}\nResumo: {wiki_page.summary}")
        except:
            pass
    if not resumos:
        return "Busca não teve retorno"
    else:
        return "\n\n".join(resumos)
```

```python
busca_wikipedia.invoke("langchain")
```

```python

import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
```

```python
# Carregar variáveis de ambiente
load_dotenv()
chat = ChatOpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key = os.getenv("ROUTER_API"),
    model="gpt-oss-120b:free", 
    temperature=0.7
)
```

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", "Você é um assistente que sabe sobre futebol"),
    ("user", "{input}")
])

tools = [busca_wikipedia]

chain = prompt | chat.bind_tools(tools)

resposta = chain.invoke({"input": "olá"})
print(resposta)
```

```python
respost = chain.invoke({"input": "Quem foi melhor Ronaldinho Gaúcho ou Messi"})
print(respost)
```
